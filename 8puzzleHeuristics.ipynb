```python
def misplaced_tiles(state, goal):
    count = 0
    for i in range(len(state)):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1
    return count

def manhattan_distance(state, goal):
    distance = 0
    for i in range(len(state)):
        val = state[i]
        if val != 0:
            current_row, current_col = divmod(i, 3)
            goal_idx = goal.index(val)
            goal_row, goal_col = divmod(goal_idx, 3)
            distance += abs(current_row - goal_row) + abs(current_col - goal_col)
    return distance

state = [2,8, 3,
         1,6,4,
         7,0,5]


goal = [1,2,3,
        8,0,4,
        7,6,5]

print("Misplaced Tiles:", misplaced_tiles(state, goal))
print("Manhattan Distance:", manhattan_distance(state, goal))
```

```python
from queue import PriorityQueue

def get_neighbors(state):
    neighbors = []
    zero_idx = state.index(0)
    r, c = divmod(zero_idx, 3)
    moves = []
    if r > 0: moves.append(-3)
    if r < 2: moves.append(3)
    if c > 0: moves.append(-1)
    if c < 2: moves.append(1)
    for move in moves:
        new_state = list(state)
        new_state[zero_idx], new_state[zero_idx + move] = new_state[zero_idx + move], new_state[zero_idx]
        neighbors.append(tuple(new_state))
    return neighbors

def a_star(start, goal, heuristic):
    start = tuple(start)
    goal = tuple(goal)
    pq = PriorityQueue()
    pq.put((heuristic(start, goal), 0, start, [start]))
    visited = {start: 0}
    while not pq.empty():
        _, cost, current, path = pq.get()
        if current == goal:
            return path
        for neighbor in get_neighbors(current):
            new_cost = cost + 1
            if neighbor not in visited or new_cost < visited[neighbor]:
                visited[neighbor] = new_cost
                priority = new_cost + heuristic(neighbor, goal)
                pq.put((priority, new_cost, neighbor, path + [neighbor]))
    return None

def print_steps(path):
    for step, state in enumerate(path):
        print(f"Step {step}:")
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()

print("A* with Misplaced Tiles")
path_misplaced = a_star(state, goal, misplaced_tiles)
print_steps(path_misplaced)

print("A* with Manhattan Distance ")
path_manhattan = a_star(state, goal, manhattan_distance)
print_steps(path_manhattan)
```
