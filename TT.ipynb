```python
class Board:
  def __init__(self,player1,player2):
    self.player1=player1
    self.player2=player2
    self.board=[[-1]*3 for _ in range(3)]

  def place(self,player,row,col):
    p=0
    if self.board[row][col]!=-1:
      print('position occupied. try again\n')
    else:
      self.board[row][col]=player.player
      p=1
    player.rows[row]+=1
    player.columns[col]+=1
    if (row==col): player.diag[0]+=1
    if (row==2-col): player.diag[1]+=1


  def check(self, player):

    for i in range(3):
      if (player.rows[i]==3 or  player.columns[i]==3): return True
    if player.diag[0]==3 or player.diag[1]==3: return True
    return False

  def print(self):
    for i in range(3):
      for j in range(3):
        if (self.board[i][j]==-1): print('-',end='')
        else: print(self.board[i][j],end='')
      print()

  def t_input(self):
    row=int(input('Enter row (0-2)'))
    column=int(input('Enter Column (0-2)'))
    return row,column

  def play(self):
    p1=self.player1
    p2=self.player2
    for i in range(9):
      self.print()
      if (i%2==0):
        print(p1.player, ' turn \n')
        row,column=self.t_input()
        self.place(p1,row,column)
      else:
        print(p2.player, 'turn \n')
        row,column=self.t_input()
        self.place(p2,row,column)

      if self.check(p1):
        print(p1.player, "won\n")
        return 0
      elif self.check(p2):
        print(p2.player, "won\n")
        return 0

    print("No one won.")



class Player:
  def __init__(self, player):
    self.player=str(player)
    self.rows=[0]*3
    self.columns=[0]*3
    self.diag=[0]*2
```

```python
krish=Player('X')
razan=Player('O')



b1=Board(krish,razan)
b1.play()
```
