```python
#Vaccum cleaner is a 3 tuple
s=[0]*3

s[1]=1 #0 means clean, 1 means dirty.
s[2]=1
def clean(i):
  s[i]=0

def moveLeft():
  s[0]=0

def moveRight():
  s[0]=1

#algorithm
while (sum(s[1:])):
  print("the floor is dirty: ",s)
  if not s[0]:
    if (s[1]):
      clean(1)
    elif (s[2]):
      moveRight()
      clean(2)
  else:
    if (s[2]):
      clean(2)
    elif (s[1]):
      moveLeft()
      clean(1)


print("The floor is clean.",s)
```

```python
#8 puzzle

import collections

#initial grid config:
gridi=[1,2,3,
       4,0,6,
       7,5,8]

gridf=[1,2,3,
       4,5,6,
       7,8,0]


def printgrid(grid):
    print(grid[:3])
    print(grid[3:6])
    print(grid[6:])
    print()
def ltoT(grid_list):
    return tuple(grid_list)

def find_blank(grid):
    return grid.index(0)

def getNbrs(grid):
    nbrs = []
    bI = find_blank(grid)

    moves = {
        0: [1, 3], 1: [0, 2, 4],
        2: [1, 5],  3: [0, 4, 6],
        4: [1, 3, 5, 7],5: [2, 4, 8],
        6: [3, 7],7: [4, 6, 8], 8: [5, 7]
    }

    for mI in moves[bI]:
        nG = list(grid)
        nG[bI], nG[mI] = nG[mI], nG[bI]
        nbrs.append(tuple(nG))

    print()
    print("Neighbours of grid: ")
    printgrid(grid)
    print("*"*10)
    for i in nbrs:
        printgrid(i)
    print("*"*10)
    print()
    return nbrs


def bfs(igL, fgL):
    igT = ltoT(igL)
    fgT = ltoT(fgL)

    queue = collections.deque([(igT, [])])
    visited = {igT}

    while queue:
        cgT, path = queue.popleft()

        if cgT == fgT:
            return path + [cgT]

        for nT in getNbrs(list(cgT)):
            if nT not in visited:
                visited.add(nT)
                queue.append((nT, path + [cgT]))

    return None

print("initial grid: ")
printgrid(gridi)
print("final grid: ")
printgrid(gridf)

sol = bfs(gridi, gridf)
if sol:
    print("sol with path")
    for step in sol:
        printgrid(step)
else:
    print("No solution exists.")


```

```python
def dls(cgT, fgT, limit, path, visited):
    if cgT == fgT:
        return path + [cgT]

    if limit == 0:
        return None


    for NT in getNbrs(list(cgT)):
        if NT not in path:
            result = dls(NT, fgT, limit - 1, path + [cgT], visited)
            if result:
                return result
    return None

def ids(initial_grid_list, final_grid_list, max_depth=20):
    igT = ltoT(initial_grid_list)
    fgT = ltoT(final_grid_list)

    for depth in range(max_depth + 1):
        print(f"Trying depth limit: {depth}")
        seen = set()
        result = dls(igT, fgT, depth, [], seen)
        if result:
            return result
    return None

print("Running IDS with initial grid:")
printgrid(gridi)
print("and final grid:")
printgrid(gridf)

sol_ids = ids(gridi, gridf)

if sol_ids:
    print("\nSolution found by IDS!")
    for step in sol_ids:
        printgrid(step)
else:
    print("\nNo solution found within the maximum depth limit.")
```
